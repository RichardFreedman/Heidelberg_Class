## 1. Introduction to DataFrames

Pandas **DataFrames** are the basic unit upon which all operations take place. Think of them as spreadsheets: rows of records and named columns.

Pandas can import (and export) data in many formats — **CSV** (comma separated values), **Excel**, **JSON** (JavaScript Object Notation, similar to a Python dictionary), and more.

### Import the Pandas Library

By convention, Pandas is imported under the short name `pd`.

````python
import pandas as pd
````

### Selecting Rows: `iloc` and `loc`

`df.iloc` and `df.loc` are *not* the same!

#### `iloc` — select by **integer index position**

Syntax: `df.iloc[start_row : end_row, start_col : end_col]`

- Square brackets `[ ]` indicate that we are taking a *slice* (a section) of the DataFrame.
- The first number is *inclusive* and the second is *exclusive*: `iloc[10:15]` gives rows 10, 11, 12, 13, 14 — but *not* row 15.
- Omit the first number to start at the beginning; omit the last to go to the end. Omit the column range to get all columns.

````python
# Rows 10–14, all columns
beatles.iloc[10:15]
````

````python
# Rows 10–14, columns 1–3 (Title, Year, Album)
beatles.iloc[10:15, 1:4]
````

#### `loc` — select by **label**

`loc` selects rows by the *label* in the index. With the default numeric index, labels and positions look the same. `loc` really becomes useful when the index is made of strings — for instance, if we make the song `Title` the index with `set_index()`:

````python

# here we are using the Title column as the index
# this allows us to access rows by their title rather than by their integer position

# note that this does not change the original DataFrame
# it just creates a new one with the index set to the Title column
beatles_by_title = beatles.set_index('Title')

````

````python
# this returns the unique row that matches the index value 'Yesterday'
# it will look like a DataFrame, but it is actually a Series because there is only one row

beatles_by_title.loc['Yesterday']
````

````python
# Several songs at once, and only the columns we name
# first the lists
song_list = ['Help!', 'Hey Jude', 'Let It Be']
column_list = ['Year', 'Album', 'Lead vocal(s)']   

# now we pass the lists to the .loc[] method to get the rows and columns we want
# in this case 'loc' is taking two arguments: 
# the first is the list of songs, and the second is the list of columns
beatles_by_title.loc[song_list, column_list]
````

### Dropping Rows

You can remove ("drop") rows by referring to their index numbers. Here we drop rows 0, 2, and 8, which are alternate takes and studio jams.

Your DataFrame will then be missing those index numbers. Use `.reset_index(drop=True)` to make the index continuous again (0 to the end).

````python
beatles_trimmed = beatles.drop([0, 2, 8])
beatles_trimmed = beatles_trimmed.reset_index(drop=True)
beatles_trimmed.head(10)
````

### Add a Column

Assign a new column name to an expression that is evaluated for every row.

Here we create a column `'sad'`. If a song's Spotify `Valence` (musical "positiveness") is less than 0.2, `'sad'` will be `True`; otherwise it will be `False`.

NB:  later we will learn how to use **fitler** the results so we see only the sad songs.

````python
beatles['sad'] = beatles['Valence'] < 0.2
beatles[['Title', 'Valence', 'sad']].head(10)
````

### Drop a Column

Columns to drop must be passed as a **list**, even if there is only one. Here we drop the Spotify `URI` column, which we won't need.

NB:  if you have already dropped a column, you may need to restart the kernel and re-run the notebook to get it back.  And if you rerun the same cell after you have dropped the column, you will get an error because the column is no longer there!

````python


beatles = beatles.drop(columns=['URI']) # notice that the 'URI' column name is square brackets!
beatles.columns
````

### Rename Columns

Several column names in this dataset contain spaces and punctuation, like `'Album debut'` and `'Songwriter(s)'`. They work, but they are awkward to type. It is good practice to use short names without spaces or special characters.

**Option 1** — copy-then-drop (useful for a single column):

````python
# Copy 'Album debut' to a new 'album_debut' column, then drop the original
beatles['album_debut'] = beatles['Album debut']

# here we drop the original 'Album debut' column, which we no longer need
beatles = beatles.drop(columns=['Album debut'])
beatles.columns
````

**Option 2** — rename with a dictionary (cleaner for several columns at once). Keys are the old names, and values are the new ones:

````python
# dictionary has the OLD column names as KEYS and the NEW column names as VALUES

col_dict = {
    'Songwriter(s)': 'songwriter',
    'Lead vocal(s)': 'lead_vocal',
    'Top 50 (Billboard)': 'top_50_billboard'
}

# now we pass the dictionary to the rename() method to change the column names

beatles = beatles.rename(columns=col_dict)
beatles.columns
````

**Tip: `dict.fromkeys()` as a shortcut**

`dict.fromkeys()` makes a dictionary instantly, with the current column names as the keys and `None` as each value. Fill in only the names you want to change and delete the rest, then finish with `df.rename(columns=col_dict)`. This is handy when there are many columns.

NB:  You will ened to copy the output of col_dict into a cell in order to edit it.  You cannot edit the output of a cell directly.

````python
col_dict = dict.fromkeys(beatles.columns)
col_dict
````

### A Column is a Series

An individual column is called a **Series**, which has its own methods.

| Expression | What it returns |
|------------|----------------|
| `df["col"]` | The column as a Series |
| `df["col"].unique()` | Array of the unique values |
| `df["col"].nunique()` | *Number* of unique values |
| `df["col"].value_counts()` | Count of each unique value |

> You can also reference a column like this: `beatles.Album`. But this "dot" syntax does not work for column names with spaces or special characters, such as `beatles["Single A side"]`. That is another good reason to rename such columns. The examples here use the bracketed syntax.

````python

# one column as a Series
beatles["Album"]
````

````python
# the unique values in that series
beatles["Album"].unique()
````

````python
# how MANY unique values in that series
beatles["Album"].nunique()
````

````python
# the count of each unique value in that series 
beatles["Album"].value_counts()
````

## 4. Sort and Count

Pandas has built-in methods for sorting and summarizing data, so you don't need to write any loops. Many more are listed on the [Pandas Cheat Sheet](https://pandas.pydata.org/Pandas_Cheat_Sheet.pdf).

### Sort Values

`sort_values()` sorts by any column. 

The order is **ascending** (alphabetical or numerical) **by default**. Reverse it with `ascending=False`.

````python
# Least danceable songs first
selected_columns = ['Title', 'Album', 'Danceability']
beatles.sort_values("Danceability")[selected_columns]
````

````python
# Longest songs first (Duration is in seconds)
selected_columns = ['Title', 'Album', 'Duration']       
beatles.sort_values("Duration", ascending=False)[selected_columns].head(10)
````

````python
# The Billboard Top 50, from number 1 down
selected_columns = ['Title', 'Year', 'top_50_billboard']    
beatles.sort_values("top_50_billboard")[selected_columns].head(10)
````

### Count Values

We saw this above, but worth repeating:

`value_counts()` returns the number of entries for each unique value in a column, from most to least frequent.

````python
beatles["Album"].value_counts()
````

````python
beatles["lead_vocal"].value_counts()
````

You may want to store the result of `value_counts()` in a new DataFrame:

````python
album_counts = pd.DataFrame(beatles["Album"].value_counts())
album_counts
````

## 5. Combining and Merging DataFrames

When you work with several datasets, you often need to combine them. There are two main ways to do that:

| Operation | When to use |
|-----------|-------------|
| **`pd.concat()`** | Stack DataFrames that share the **same columns** (add more rows) |
| **`pd.merge()`** | Join DataFrames that share a **key column** but have different columns (add new columns) |

**Joining** (`df.join()`) is similar to merging but less versatile, since it combines frames on their index only.

Read more: [concat](https://pandas.pydata.org/docs/reference/api/pandas.concat.html) · [merge](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html) · [join](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html)

### Concatenation vs. Merging

- If your DataFrames have **identical column names**, you *concatenate*.
- If they have different numbers of columns but otherwise share column names, you *might* be able to concatenate, as long as you don't mind lots of missing data in some places.
- If your DataFrames are quite different but **share some common data** (like song titles), you *merge*.

The Beatles CSV in this notebook has *already* been assembled this way: it is the Spotify data merged with the University of Belgrade data. So we don't need to combine anything here. But the syntax is worth knowing for when you have separate sources of your own.

### A Hypothetical Example

Suppose you had two separate DataFrames: `spotify_df` (with a `song` column) and `billboard_df` (with a `Title` column). The general pattern would look like this. *(This is shown for reference only, not run.)*

```python
# Concatenate: stack frames with the same columns
all_songs = pd.concat([songs_part_1, songs_part_2]).reset_index(drop=True)

# Merge: join frames that share a key column
combined = pd.merge(left=billboard_df,
                    right=spotify_df,
                    left_on="Title",
                    right_on="song",
                    how="inner")
```

Key arguments to `pd.merge()`:
- The two frames to be joined are called `left` and `right`.
- `left_on` and `right_on` name the shared column in each frame. The columns don't need the same name, but they must contain the same items (like song titles). If the names *are* the same, you can just write `on="Title"`.
- `how` says which keys to keep. With `"inner"`, only rows with a match in **both** frames are kept. With `"left"`, every row of the left frame is kept, and missing values are filled with `NaN`.
- The optional `suffixes` argument tells Pandas how to label columns that have the same name in both frames.

**One caution:** Pandas only matches keys that are formatted in *exactly* the same way. `"12-Bar Original"` and `"12-bar Original"` count as different songs. So the key columns usually need to be cleaned first.
