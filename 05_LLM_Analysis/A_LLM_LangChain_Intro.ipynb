#  LLM with Langchain

In this notebook you will explore the very basics of working with LLMs with the help of Langchain, one of several Python libraries that can manage the choice of model, API keys, and prompts.

Langchain is a Python framework that allows us to work with _any_ LLM, including OpenAI's GPT models, Cohere, and others. It provides a simple interface to interact with these models, manage prompts, and handle responses.

## 1c.  Select LLM Model

Now, you can select the model you want to use. For this course, we will use models from OpenAI. Here we'll use gpt-4o-mini because it's cheap and actually quite capable. Newer models like gpt-5o-mini would work too (maybe even better), but we're sticking with the model we've used in the past for now.

```python
# Select model
from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-4o-mini", model_provider="openai")
```

## 2a.  A First Prompt

Now, let's see if it works with a basic prompt.

```python
print(llm.invoke("Knock knock").content)
```

## 2b.  A System Prompt

Now that we have API e can create our own custom chat model by using what's called a **system prompt**. 

The system prompt basically tells the LLM what it's job is. For the ChatGPT website, it might be something like "You are a helpful model, don't violate any content guidelines..."  

Here's how we can write our own. 

### Talk like a Pirate?

Here we will create a system prompt that tells the model to talk like a pirate.  You could adapt this to make the model talk like a lawyer, a doctor, a baby, or even a specific person.  The possibilities are endless.

```python
from langchain_core.prompts import ChatPromptTemplate

# here we write our prompt.  
prompt = ChatPromptTemplate.from_messages([
    ("system", """You are the captain of a pirate ship. You may only respond to questions like a pirate. 
    Some examples of common pirate phrases are 'Ahoy Matey', 'Aye Aye Captain,' 'Shiver me timbers,' etc."""),
    ("human", "Context:\n{context}\n\nQuestion: {question}") # Keep the rest of this the same. Just replace the system message about pirates, and you'll have your own.
])
```

Now we'll set up a simple function so that you can easily ask your own questions.

```python
def ask_llm(question:str):
    message = prompt.invoke({"question": question, "context": ""}) # Context is blank for now, but later it'll be the documents you're working with for RAG.
    response = llm.invoke(message)
    print(response.content)
```

#### Test 1:  A User Prompt

Now you can ask the model a question and see how it responds. 


```python
ask_llm("Why don't you tell me a story.")
```
